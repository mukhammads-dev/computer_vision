```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/cat_image.webp")

cv2_imshow(image)
```

```python
print("SHAPE:", image.shape)
# 800 => height
# 800 => width
# 3 => color channels (BGR) => Blue, Green, Red
print("HEIGHT:", image.shape[0])
print("WIDTH:", image.shape[1])
print("CHANNELS:", image.shape[2])
```

```python
(b, g, r) = image[100, 50] # 100chi qatar / 50chi ustun ranglarini olib beradi
print("blue:", b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) # rangni ozgartirish uchun cvtColor
cv2_imshow(gray)
```

```python
# rasmni olchami bilan ishlash
resized = cv2.resize(gray, (300, 300))
cv2_imshow(resized)
```

```python
cropped = gray[100:300, 150:350]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE) # 90, 180 gradusga ogirib beradi
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 1) # ongdan chapga ozgartirib beradi
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped_cat.jpg", flipped) # file saqlash uchun
```
